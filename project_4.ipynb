```python
# ==============================
# Data Cleaning & Reporting Workflow
# ==============================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Load dataset (chunking for large files)
chunk_size = 100000
chunks = pd.read_csv("train.csv", chunksize=chunk_size)
df = pd.concat(chunks, ignore_index=True)

print("Initial Shape:", df.shape)
print("Columns:", df.columns.tolist())

# Step 2: Handle missing values
# Numeric columns → fill with median (robust against outliers)
for col in df.select_dtypes(include=['float64','int64']).columns:
    df[col] = df[col].fillna(df[col].median())

# Categorical columns → fill with mode (most frequent value)
for col in df.select_dtypes(include=['object','string']).columns:
    df[col] = df[col].fillna(df[col].mode()[0])

# Step 3: Standardize text columns
for col in df.select_dtypes(include=['object','string']).columns:
    df[col] = df[col].str.strip().str.lower()

# Step 4: Handle duplicates
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)

df = df.drop_duplicates()

# Step 5: Automated reporting
summary = df.describe(include='all')
print(summary)

# Example grouped report (replace with actual column names from your dataset)
# For demonstration, let's assume you have 'Product' and 'Sales' columns
if 'Product' in df.columns and 'Sales' in df.columns:
    report = df.groupby('Product')['Sales'].sum().reset_index()
    print(report)

# Step 6: Visual summaries
# Bar chart (replace with actual categorical + numeric columns)
if 'Product' in df.columns and 'Sales' in df.columns:
    sns.barplot(x="Product", y="Sales", data=df, estimator=sum)
    plt.title("Total Sales by Product")
    plt.xticks(rotation=45)
    plt.show()

# Trend line (replace with actual date + numeric columns)
if 'Date' in df.columns and 'Sales' in df.columns:
    df.groupby('Date')['Sales'].sum().plot(kind='line', figsize=(10,5))
    plt.title("Sales Trend Over Time")
    plt.ylabel("Total Sales")
    plt.show()

# Step 7: Export cleaned data & reports
df.to_csv("cleaned_train.csv", index=False)

if 'Product' in df.columns and 'Sales' in df.columns:
    report.to_csv("sales_report.csv", index=False)

print("✅ Workflow complete: cleaned")
```

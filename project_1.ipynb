```python
import pandas as pd
import datetime

# Load dataset
df = pd.read_csv("Chocolate_Sales.csv")

# Clean columns
df["Date"] = pd.to_datetime(df["Date"])
df["Amount"] = df["Amount"].replace('[\$,]', '', regex=True).astype(float)
df["Boxes Shipped"] = pd.to_numeric(df["Boxes Shipped"], errors="coerce")

print(df.head())

#Top products by revenue 
top_products = df.groupby("Product")["Amount"].sum().sort_values(ascending=False).head(5)
print(top_products)

#Top countries by revenue
top_countries = df.groupby("Country")["Amount"].sum().sort_values(ascending=False).head(5)
print(top_countries)

import matplotlib.pyplot as plt
import seaborn as sns

# Revenue trend over time
plt.figure(figsize=(12,6))
sns.lineplot(x="Date", y="Amount", data=df)
plt.title("Revenue Trend Over Time")
plt.show()

# Top products
top_products.plot(kind="bar", title="Top 5 Products by Revenue", figsize=(8,5))
plt.show()

# Top countries
top_countries.plot(kind="bar", title="Top 5 Countries by Revenue", figsize=(8,5))
plt.show()



```
